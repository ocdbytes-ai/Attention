```python
import torch
from torch import nn
```

```python
# Shape : 2 x 3 x 4
# 3D Tensor
a = torch.rand(2, 3, 4)
a.size(1)
a
```

```python
# dim == 0
# along dimension 0 --> a.size(0) == 2
# un fix dimension 0 and apply softmax for:
# a[0, 0 ,0]
# a[1, 0 ,0]
nn.functional.softmax(a, dim=0)
```

```python
# dim == 1
# along dimension 1 --> a.size(0) == 3
# un fix dimension 1 and apply softmax for:
# a[0, 0 ,0]
# a[0, 1 ,0]
# a[0, 2 ,0]
nn.functional.softmax(a, dim=1)
```

```python
# dim == 2
# along dimension 2 --> a.size(0) == 4
# un fix dimension 2 and apply softmax for:
# a[0, 0 ,0]
# a[0, 0 ,1]
# a[0, 0 ,2]
# a[0, 0 ,3]
nn.functional.softmax(a, dim=2)
```
