# RNN and Attention

## Model Description 

We are describing the model in the same way as encoder decoder architecture for language translation as done before but here along with the original text (encoder hidden state) passed to decoder we also send the text which was already generated for next token generation. 

$$
\mathbf{c}_{t'} = \sum_{t=1}^{T} \alpha(\mathbf{s}_{t' - 1}, \mathbf{h}_{t}) \mathbf{h}_{t}
$$

This yields $c_{t'}$, which is updated after any decoding time step $t'$. Suppose that the input sequence is of length $T$. We used $s_{t'-1}$ as the query, and $h_t$ as both the key and the value. Note that $c_{t'}$ is then used to generate the state $s_{t'}$ and to generate a new token.
